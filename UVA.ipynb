```python
# Importamos librerías
import pandas as pd
import numpy as np
```

```python
# Carga del dataset
URL = "https://raw.githubusercontent.com/fnlopezsanz/TP_Lab_Met_Cuantitativos_Inversiones/refs/heads/main/"

df_uva = pd.read_csv(URL + "unidad-de-valor-adquisitivo-uva-20250101-20251231.csv")

# Análisis e inspección inicial de la estructura
print("Dimensiones:", df_uva.shape)
print("El dataset posee 365 registros y 2 variables. Cada registro representa el valor diario de la UVA durante el año 2025.")

df_uva.head()
df_uva.tail()
df_uva.info()
```

```python
# Convertimos fecha
df_uva["Fecha"] = pd.to_datetime(df_uva["Fecha"])

# Datos iniciales
capital_inicial = 1_000_000

uva_inicial = df_uva.iloc[0]["Valor"]
uva_final = df_uva.iloc[-1]["Valor"]

# Cantidad de UVA compradas
cantidad_uva = capital_inicial / uva_inicial

# Capital final
capital_final = cantidad_uva * uva_final

# Rendimiento
rendimiento_pct = (capital_final / capital_inicial - 1) * 100

print(f"UVA inicial: ${uva_inicial:.2f}")
print(f"UVA final: ${uva_final:.2f}")
print(f"Cantidad de UVA compradas con $1.000.000: {cantidad_uva:.2f}")
print(f"Capital final: ${capital_final:,.2f}")
print(f"Rendimiento: {rendimiento_pct:.2f}%")
```
